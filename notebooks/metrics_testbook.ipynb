Data Conversion and Metric Calculation

```python
import sys
from pathlib import Path

sys.path.append(str(Path.cwd().parent)) # Project folder

from data_utils import load_companies, load_data, find_project_root

data_path = find_project_root() / "data" / "AssetDB.csv"
companies, df = load_companies(data_path)
print(len(companies), "assets loaded")
```

```python
assetdata = load_data(companies) # Load past 50-day test window
```

```python
prices = assetdata["Close"] # one column per ticker, one row per trading day
prices = prices.ffill()
prices.tail()
```

```python
returns = prices.pct_change().dropna(how="all")
returns.tail()
```
